## Mean square error gd
### Linear regression using gradient descent

```python
from helper import *
```

```python
def mean_square_error_gd(y, tx, initial_w, max_iters, gamma):
    """
    Computes linear regression using gradient descent
    
    Args:
        y:  shape=(N, )
        tx:  shape=(N, )
        initial_w: shape=(2, ) initial weight vector ; shape=
        max_iters: number of steps to run ; int
        gamma: stepsize ; int

    Return:
        (w, loss): last weight vector of the method, corresponding loss value 
    """
    #define intial parameters
    w = initial_w
    
    # iterate through the steps to run and change w
    for n_iter in range(max_iters):
        grad, _ = compute_mse_gradient(y, tx, w)
        w = w - gamma * grad

    #the w that comes out of the loop is the "last one", so we can compute loss from it.
    _, err = compute_mse_gradient(y, tx, w)
    loss = compute_mse(err)
        
    return (w, loss)
```

## Mean square error sgd
### Linear regression using stochastic gradient descent

```python
def mean_square_error_sgd(y, tx, initial_w, max_iters, gamma, batch_size = 1):
    """
    Computes linear regression using stochastic gradient descent
    
    Args:
        y:  shape=(N, )
        tx:  shape=(N, )
        initial_w: shape=(2, ) initial weight vector ; shape=
        max_iters: number of steps to run ; int
        gamma: stepsize ; int

    Return:
        (w, loss): last weight vector of the method, corresponding loss value 
    """
    #define initial parameters
    w = initial_w
    #iterate through the steps to run and change w
    for n_iter in range(max_iters):
        
        #iterate through the different stochastic batches (only 1 as asked in the énoncé)
        for y_batch, tx_batch in batch_iter(y, tx, batch_size=batch_size, num_batches=1):
            grad, _ = compute_mse_gradient(y_batch, tx_batch, w)
            w = w - gamma * grad

    #the w that comes out of both loops is the "last one", so we can compute loss from it.
    _, err = compute_mse_gradient(y, tx, w)
    loss = compute_mse(err)
        
    return (w, loss)
    
```

## Least squares
### Least squares regression using normal equations

```python
def least_squares(y, tx):
    """
    Calculate the least squares solution.

    Args:
        y: shape=(N, ), N is the number of samples.
        tx: shape=(N, )

    Returns:
        (w, loss): last weight vector of the method, corresponding loss value 

    """
    
    a = tx.T.dot(tx)
    b = tx.T.dot(y)
    w = np.linalg.solve(a, b)
    
    loss = compute_mse_loss(y, tx, w)
    
    return (w, loss)
```

## Ridge Regression
### Ridge regression using normal equations

```python
def ridge_regression(y, tx, lambda_):
    """
    Implement ridge regression.

    Args:
        y: shape=(N,), N is the number of samples.
        tx: shape= (N,),
        lambda_: scalling ; float

    Returns:
        (w,loss): last weight vector of the method, corresponding loss value 

    """
    
    aI = 2 * tx.shape[0] * lambda_ * np.identity(tx.shape[1])
    a = tx.T.dot(tx) + aI
    b = tx.T.dot(y)
    
    w = np.linalg.solve(a, b)
    loss = compute_mse_loss(y, tx, w)
    
    return(w, loss)
```
